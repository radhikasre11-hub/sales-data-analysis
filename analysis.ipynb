```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df["ORDERDATE"] = pd.to_datetime(df["ORDERDATE"])

sns.set_theme(style="whitegrid")
```

```python
df.groupby("PRODUCTLINE")["SALES"].sum().sort_values().plot(
    kind="bar", figsize=(9, 5), color="steelblue"
)
plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df.groupby("TERRITORY")["SALES"].sum().sort_values().plot(
    kind="bar", figsize=(8, 5), color="teal"
)
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
monthly_sales = df.set_index("ORDERDATE")["SALES"].resample("MS").sum()

monthly_sales.plot(figsize=(10, 5), marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
top_products = df.groupby("PRODUCTCODE")["SALES"].sum().nlargest(10)

top_products.sort_values().plot(
    kind="barh", figsize=(9, 5), color="coral"
)
plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product Code")
plt.tight_layout()
plt.show()
```

```python
df.groupby("DEALSIZE")["SALES"].sum().plot(
    kind="pie", autopct="%1.1f%%", figsize=(7, 7)
)
plt.title("Sales Distribution by Deal Size")
plt.ylabel("")
plt.show()
```

```python
top_countries = df.groupby("COUNTRY")["SALES"].sum().nlargest(10)

top_countries.sort_values().plot(
    kind="barh", figsize=(9, 5), color="mediumpurple"
)
plt.title("Top 10 Countries by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

```python
print("Total Revenue:", round(df["SALES"].sum(), 2))
print("Average Sale:", round(df["SALES"].mean(), 2))
print("Total Orders:", df["ORDERNUMBER"].nunique())
print("Best-Selling Category:",
      df.groupby("PRODUCTLINE")["SALES"].sum().idxmax())
print("Top Region:",
      df.groupby("TERRITORY")["SALES"].sum().idxmax())
print("Top Country:",
      df.groupby("COUNTRY")["SALES"].sum().idxmax())
```
